#  5. Convert Date Columns to Datetime

```python
# Convert all date columns to proper datetime format.
# Convert all date columns to proper datetime format
users['signup_date'] = pd.to_datetime(users['signup_date'])
payments['payment_date'] = pd.to_datetime(payments['payment_date'])
ratings['rating_date'] = pd.to_datetime(ratings['rating_date'])
viewership['watch_date'] = pd.to_datetime(viewership['watch_date'])

# This column is in DD-MM-YYYY format, so we tell pandas explicitly
titles['content_added_date'] = pd.to_datetime(titles['content_added_date'], format='%d-%m-%Y')

```

# 6. Data Cleaning

```python
# Clean Users Data
```

```python
# Step 5.1: Fill missing age with median age
users['age'] = users['age'].fillna(users['age'].median())

# Fill missing gender with 'Not Specified'
users['gender'] = users['gender'].fillna('Not Specified')

# Fill missing primary_device with most common device
users['primary_device'] = users['primary_device'].fillna(users['primary_device'].mode()[0])

```

```python
#Check again
users.isnull().sum()
```

```python
# Clean Payments Data
```

```python
# Fill missing payment_method with 'Unknown'.
payments['payment_method'] = payments['payment_method'].fillna('Unknown')
```

```python
# Check again
payments.isnull().sum()
```

```python
# Clean Ratings data
```

```python
# Fill missing 'review_text' with 'No review given'.
ratings['review_text'] = ratings['review_text'].fillna('No review given')
```

```python
# Check again
ratings.isnull().sum()
```

```python
# Clean viewership data
```

```python
# Step 5.5: Fill missing watch_minutes with median
viewership['watch_minutes'] = viewership['watch_minutes'].fillna(viewership['watch_minutes'].median())

# Fill missing completion_pct with median
viewership['completion_pct'] = viewership['completion_pct'].fillna(viewership['completion_pct'].median())

# Fill missing device with 'Unknown'
viewership['device'] = viewership['device'].fillna('Unknown')

```

```python
#Check again
viewership.isnull().sum()
```

```python
# Clean Titles data
```

```python
# Check which title_type has missing duration
titles[titles['duration_minutes'].isnull()]['title_type'].value_counts()
```

```python
# Fill duration_minutes only for Movies (TV Shows genuinely don't have a single runtime)
titles.loc[titles['title_type'] == 'Movie', 'duration_minutes'] = (
    titles.loc[titles['title_type'] == 'Movie', 'duration_minutes']
    .fillna(titles.loc[titles['title_type'] == 'Movie', 'duration_minutes'].median())
)
```

```python
# Check again
titles.isnull().sum()
```

# 6.1. Merge Datasets

```python
# Step 1: Calculate total amount spent by each user
total_spent_per_user = payments.groupby('user_id')['final_amount'].sum()

# Step 2: Calculate number of payments made by each user
payment_count_per_user = payments.groupby('user_id')['payment_id'].count()

# Step 3: Merge both into the users table (left join keeps all users)
users = users.merge(total_spent_per_user, on='user_id', how='left')
users = users.rename(columns={'final_amount': 'total_spent'})

users = users.merge(payment_count_per_user, on='user_id', how='left')
users = users.rename(columns={'payment_id': 'num_payments'})

# Step 4: Users with no payment history will show blank - fill with 0
users['total_spent'] = users['total_spent'].fillna(0)
users['num_payments'] = users['num_payments'].fillna(0)

users.head()
```

# 6.2. Feature Engineering

```python
# Step 1: Create age groups from the age column
users['age_group'] = pd.cut(users['age'], bins=[0, 18, 25, 35, 50, 100],
                              labels=['Under 18', '18-25', '26-35', '36-50', '50+'])

# Step 2: Make sure signup_date is in proper datetime format
users['signup_date'] = pd.to_datetime(users['signup_date'])

# Step 3: Calculate how many days ago the user signed up
today = pd.Timestamp('today')
users['tenure_days'] = (today - users['signup_date']).dt.days

# Check the new columns
users[['age', 'age_group', 'signup_date', 'tenure_days']].head()
```

# 7. Business Question and Insights

```python
### Q1. What is the total revenue generated, and which subscription plan contributes the most?
```

```python
# Q1: Total Revenue & Revenue by Plan
successful_payments = payments[payments['payment_status'] == 'Success']
revenue_by_plan = successful_payments.groupby('plan')['final_amount'].sum().sort_values(ascending=False)

print("Total Revenue: ₹", round(successful_payments['final_amount'].sum(), 2))
print(revenue_by_plan)

# Chart - Netflix style colors (red shades)
plt.figure(figsize=(8,5))
sns.barplot(x=revenue_by_plan.index, y=revenue_by_plan.values, color="#E50914")
plt.title("Revenue by Subscription Plan")
plt.xlabel("Plan")
plt.ylabel("Total Revenue (₹)")
plt.show()
```

### INSIGHTS 


The platform generated a total revenue of ₹22,15,354. The Standard plan is the biggest revenue driver, contributing more than any other plan, followed by Premium. This suggests most users prefer a mid-tier plan rather than the cheapest (Mobile) or the most expensive (Premium) option.

```python
### Q2. Which countries have the highest number of users?
```

```python
# Q2: Top 5 countries by number of users
top_countries = users['country'].value_counts().head(5)
print(top_countries)

# Chart - horizontal bar, Netflix red to dark gradient
plt.figure(figsize=(8,5))
sns.barplot(x=top_countries.values, y=top_countries.index, color="darkred")
plt.title("Top 5 Countries by User Count")
plt.xlabel("Number of Users")
plt.ylabel("Country")
plt.show()
```

## Insight:

India has the highest number of users, showing that the platform's largest market is domestic. Other major markets include the USA, UK, Canada, and Japan, indicating a strong international user base as well.

```python
### Q3. What proportion of users are currently active vs inactive?
```

```python
# Q3: Active vs Inactive users
status_counts = users['is_active'].value_counts()
status_counts.index = ['Active', 'Inactive']
print(status_counts)

# Chart
plt.figure(figsize=(6,6))
plt.pie(status_counts.values, labels=status_counts.index, 
         colors=["crimson", "dimgray"], startangle=90)
plt.title("Active vs Inactive Users")
plt.show()
```

```python
### Q6. How has monthly revenue trended over time?
```

```python
# Q6: Monthly revenue trend
successful_payments = payments[payments['payment_status'] == 'Success'].copy()
successful_payments['payment_month'] = successful_payments['payment_date'].dt.to_period('M')
monthly_revenue = successful_payments.groupby('payment_month')['final_amount'].sum()
print(monthly_revenue)

# Chart
plt.figure(figsize=(10,5))
monthly_revenue.plot(kind='line', marker='o', color="crimson")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=45)
plt.show()
```

**Insight:**

Revenue fluctuates between ₹60,000 and ₹90,000 each month, with no steady upward trend. There is a noticeable spike around July 2025, possibly due to a promotion or new content launch. The sharp drop in the last month is likely because that month's data is still incomplete.

# 8. Additional Analysis

```python
### 7.1 Correlation Between Watch Time and Completion Percentage
```

```python
# Check correlation between numerical columns in viewership
correlation_data = viewership[['watch_minutes', 'completion_pct', 'session_count', 'buffering_events']]
correlation_matrix = correlation_data.corr()
print(correlation_matrix)

# Heatmap
plt.figure(figsize=(7,5))
sns.heatmap(correlation_matrix, annot=True, cmap="Reds")
plt.title("Correlation Between Viewing Metrics")
plt.show()
```

# Insights

Watch minutes and completion percentage show a positive correlation, meaning users who watch longer tend to complete more of the content. Buffering events show little to no strong relationship with completion, suggesting buffering isn't a major reason users stop watching.

```python
### 7.2 Average Monthly Fee by Country
```

```python
# Pivot table: average monthly fee by country
avg_fee_by_country = users.groupby('country')['monthly_fee'].mean().sort_values(ascending=False)
print(avg_fee_by_country)

# Chart
plt.figure(figsize=(8,5))
sns.barplot(x=avg_fee_by_country.values, y=avg_fee_by_country.index, color="crimson")
plt.title("Average Monthly Fee by Country")
plt.xlabel("Average Monthly Fee (₹)")
plt.ylabel("Country")
plt.show()
```
